# 02. Exploratory Data Analysis (2024?2026 Ujjain Onion Mandi)
Analyzing historical wholesale price trajectories, 2026 price movements, seasonality, and volume arrivals.

```python
import sys, os, pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
sys.path.insert(0, os.path.abspath('..'))
from src.preprocessing import load_and_preprocess

df = load_and_preprocess(save_processed=False)
df['date'] = pd.to_datetime(df['date'])
print(f'Total auction trading days: {len(df)}')
print(df[['min_price', 'max_price', 'modal_price', 'arrivals_tonnes']].describe())

plt.figure(figsize=(14, 5))
plt.plot(df['date'], df['modal_price'], color='#d9534f', linewidth=1.8, label='Daily Modal Price (Rs./Q)')
plt.plot(df['date'], df['modal_price'].rolling(7, min_periods=1).mean(), color='#0275d8', label='7-Day MA')
plt.title('Ujjain APMC: Wholesale Onion Price History (2024-2026)')
plt.xlabel('Auction Date')
plt.ylabel('Modal Price (Rs. / Quintal)')
plt.legend()
plt.show()
```
